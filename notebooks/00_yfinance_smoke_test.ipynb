```python
import yfinance as yf, time
tickers = ["LEH", "BSC", "WB", "GMI", "FB", "META", "AAPL"]
for t in tickers:
    try:
        df = yf.Ticker(t).history(period="max", auto_adjust=False)
        if df.empty:
            print(f"{t}: EMPTY")
        else:
            print(f"{t}: {len(df)} rows, {df.index.min().date()} to {df.index.max().date()}")
    except Exception as e:
        print(f"{t}: ERROR — {e}")
    time.sleep(1.5)
```

Takeaways:
- Companies that went bankrupt (LEH, BSC, GMI)are removed from yfinance --> survivorship bias.
- Ticker are reused, so ticker data may be inaccurate. For example, WB was Wachovia until 2008, then WB become Weibo Corp. 
- Going forward, we will backtest on the price data yfinance has and accept the fact that some tickers will silently drop. We will measure this bias by counting how many name-months we lose and report them. This will likely cause our Sharpe to be biased upwards, but the methodology should be correct. 
- Additionally, we will add a strict ticker-reuse check. That is, for every ticker in our universe, we verify that yfinance covers the time period we need; for example, if Wachovia was in the S&P 500 in 2002 but yfinance's "WB" ticker starts in 2014 then we will raise an error and drop the ticker. We will keep a "missing data ledger" that includes the reason for dropping tickers.
